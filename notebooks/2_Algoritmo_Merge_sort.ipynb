#Algoritmo de ordenación por fusión (merge sort)

Este algoritmo es muy popular por su eficiencia y estabilidad. Sigue el principio de "divide y vencerás". Consiste en dividir recursivamente el array de entrada en dos mitades, ordenar recursivamente cada mitad y, finalmente, volver a unirlas para obtener el array ordenado.

**¿Cómo trabaja?**
* Divide la lista o matriz recursivamente en dos mitades hasta que ya no se pueda dividir.
* Conquistar: Cada submatriz se ordena individualmente mediante el algoritmo de ordenación por fusión.
* Fusionar: Las submatrices ordenadas se vuelven a fusionar en el orden original. El proceso continúa hasta que todos los elementos de ambas submatrices se hayan fusionado.

Imagen tomada de Wikipedia, Merge sort.

```python
def merge(arr, left, mid, right):
    n1 = mid - left + 1
    n2 = right - mid

    # Create temp arrays
    L = [0] * n1
    R = [0] * n2

    # Copy data to temp arrays L[] and R[]
    for i in range(n1):
        L[i] = arr[left + i]
    for j in range(n2):
        R[j] = arr[mid + 1 + j]

    i = 0
    j = 0
    k = left

    # Merge the temp arrays back
    # into arr[left..right]
    while i < n1 and j < n2:
        if L[i] <= R[j]:
            arr[k] = L[i]
            i += 1
        else:
            arr[k] = R[j]
            j += 1
        k += 1

    # Copy the remaining elements of L[],
    # if there are any
    while i < n1:
        arr[k] = L[i]
        i += 1
        k += 1

    # Copy the remaining elements of R[],
    # if there are any
    while j < n2:
        arr[k] = R[j]
        j += 1
        k += 1

def mergeSort(arr, left, right):
    if left < right:
        mid = (left + right) // 2

        mergeSort(arr, left, mid)
        mergeSort(arr, mid + 1, right)
        merge(arr, left, mid, right)
```

```python
arr = [38, 27, 43, 10]
mergeSort(arr, 0, len(arr) - 1)
for i in arr:
  print(i, end=" ")
  print()
```

El algoritmo Merge Sort tiene una complejidad temporal de $O(n \log n)$ en todos los casos (mejor, promedio y peor).

La complejidad es exactamente la misma sin importar el estado inicial de los datos (ya sea desordenados o casi ordenados). Esto se divide en dos fases:

* División (Logarítmica): El algoritmo divide el arreglo a la mitad de forma recursiva hasta llegar a subarreglos de un solo elemento. Esto toma un tiempo logarítmico: $log_{2}(n)$.
* Fusión (Lineal): Compara y combina los subelementos ya ordenados en una sola estructura. Esta etapa toma un tiempo de orden $O(n)$ en cada nivel de la recursión.


## Ventajas y desventajas del algoritmo

**Ventajas:**

* Estabilidad: La ordenación por fusión es un algoritmo de ordenación estable, lo que significa que mantiene el orden relativo de los elementos iguales en el array de entrada.

* Rendimiento garantizado en el peor de los casos: La ordenación por fusión tiene una complejidad temporal en el peor de los casos de $O(n log n)$, lo que significa que funciona bien incluso con grandes conjuntos de datos.

* Fácil de implementar: El enfoque de divide y vencerás es sencillo.

* Paralelismo natural: La fusión de subarreglos es independiente, lo que la hace adecuada para el procesamiento paralelo.

**Desventajas:**

* Complejidad espacial: La ordenación por fusión requiere memoria adicional para almacenar los subarreglos fusionados durante el proceso de ordenación.

* No es in situ: La ordenación por fusión no es un algoritmo de ordenación in situ, lo que significa que requiere memoria adicional para almacenar los datos ordenados. Esto puede ser una desventaja en aplicaciones donde el uso de memoria es un factor importante. En general, el algoritmo Merge Sort es más lento que QuickSort, ya que QuickSort es más compatible con la caché porque funciona directamente in-situ.
